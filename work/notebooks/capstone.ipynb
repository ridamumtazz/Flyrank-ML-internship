## 2. Data

## 2. Data

This project uses the FlyRank internship warehouse dataset hosted on Hugging Face. The analysis was performed using the `fact_content_daily_performance_sample` table, which contains approximately 11.7 million daily content performance records. The data covers the period from **2025-01-27 to 2026-06-30**.

The dataset includes search and engagement signals such as Google Search Console impressions, clicks, average position, GA4 sessions, engaged sessions, and report dates. Additional features including click-through rate (CTR) and engagement rate were calculated during data preparation.

To prepare the data for modeling, the report date was converted to a datetime format, CTR and engagement rate were computed, and the data was sorted by content item and date. The prediction target was defined as the next observation's click count using a one-step forward shift within each content item. Rows with missing target values or missing feature values were excluded to ensure consistent model training.

No client names, domains, URLs, private search queries, or confidential information were used. The analysis uses only pseudonymized identifiers and public-safe aggregated performance signals.


## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

## 3. Methodology

The project follows the Refresh / Content Opportunity Scoring lane. The objective is to prioritize content items that may benefit from review based on observed search performance signals.

A Random Forest Regressor was selected because it can model non-linear relationships between multiple performance features while providing feature importance for interpretation. The model predicts the next observation's click count using historical performance data.

The feature set includes Google Search Console impressions, clicks, click-through rate (CTR), average search position, GA4 sessions, and engagement rate. The target variable is the next period's click count created using a one-step forward shift within each content item.

A time-aware split was first used to compare model performance with the baseline. An additional grouped-by-client split was then performed to evaluate generalization to unseen clients. Leakage checks confirmed that no label-derived columns, future information, product flags, or client identifiers were used as model features. Feature importance and prediction errors were also reviewed to verify that the model was learning meaningful patterns rather than memorizing the data.

The model is intended to provide decision-support for content prioritization rather than guarantee future search performance.


```python
from IPython.display import Markdown, display

display(Markdown("""

## Methodology

- Model: Random Forest Regressor
- Features: Impressions, Clicks, CTR, Average Position, Sessions, Engagement Rate
- Target: Next-period clicks
- Validation: Time-aware split and grouped-by-client split
- Leakage audit: Completed with no risky feature columns used
- Purpose: Decision-support for content prioritization

"""))
```

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

## 4. Results (vs Baseline)

The Random Forest model was evaluated using the same performance metrics across different validation strategies.

Using the time-aware split, the model achieved a Mean Absolute Error (MAE) of **0.2620** and a Root Mean Squared Error (RMSE) of **2.4575**.

To evaluate generalization, the model was tested using a grouped-by-client split. Under this more challenging validation design, the MAE increased to **0.5129** and the RMSE increased to **47.0867**. The difference between the two evaluations indicates that performance decreases when predicting for previously unseen clients.

Feature importance analysis showed that Google Search Console clicks and GA4 sessions contributed most to the model, while CTR, engagement rate, and average position had smaller contributions.

These results should be interpreted as measured model performance for content prioritization and not as evidence of causal relationships or guaranteed future outcomes.


```python
import pandas as pd

results = pd.DataFrame({
    "Validation": [
        "Time-aware split",
        "Grouped-by-client split"
    ],
    "MAE": [
        0.2620,
        0.5129
    ],
    "RMSE": [
        2.4575,
        47.0867
    ]
})

display(results)
```

## 5. Limitations

*What this work cannot claim.*

## 5. Limitations

This project is intended as a decision-support tool and has several important limitations.

The model identifies patterns in historical search and engagement data but does not establish causal relationships. A high action score does not guarantee that modifying a content item will improve future performance.

The grouped-by-client validation showed lower performance than the time-aware split, suggesting that the model generalizes less effectively to previously unseen clients. This finding supports the use of careful validation and conservative interpretation.

The recommendations are based on historical observations from the available dataset. Changes in search engine behavior, user interests, seasonality, or business priorities may reduce the usefulness of older recommendations.

All recommendations should be reviewed by a human before action is taken. The model is designed to support prioritization rather than replace expert judgment.


```python
from IPython.display import Markdown, display

display(Markdown("""

## Limitations

This work provides directional decision-support based on historical observations. Results should not be interpreted as causal or guaranteed predictions. Human review is required before implementing any recommendation.

"""))
```
